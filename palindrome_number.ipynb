# LeetCode #9 — Palindrome Number

Difficulty: Easy | [Problem Link](https://leetcode.com/problems/palindrome-number/)

## The Problem

Given an integer `x`, return `True` if `x` is a palindrome, `False` otherwise.

A **palindrome** is something that reads the same forwards and backwards.

- `121` is a palindrome (reads `121` both ways)
- `-121` is NOT a palindrome (reversed it reads `121-`)
- `10` is NOT a palindrome (reversed it reads `01`)

## Key Observations Before Coding

Before writing any code, think about what we already know:

1. **Negative numbers are never palindromes.** The `-` sign is at the front but not at the back. `-121` reversed is `121-` — not the same.

2. **Numbers ending in 0 (except 0 itself) are never palindromes.** If a number ends in `0`, its reverse would start with `0` — but no valid number starts with `0`. So `10`, `200`, `1000` are all not palindromes. The exception is `0` itself, which is a palindrome.

3. **Single-digit numbers (0-9) are always palindromes.** `7` reversed is still `7`.

````python
# Quick check — do these observations hold?
test_cases = [-121, -5, 10, 200, 0, 7, 121, 1221]

for x in test_cases:
    s = str(x)
    is_pal = s == s[::-1]
    print(f"x = {x:>5}  →  str = '{s}',  reversed = '{s[::-1]}',  palindrome = {is_pal}")
````

---
## The Solution: String Conversion Approach

````python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        s = str(x)
        return s == s[::-1]

# Test
sol = Solution()
print(sol.isPalindrome(121))    # True
print(sol.isPalindrome(-121))   # False
print(sol.isPalindrome(10))     # False
````

---
## Line-by-Line Breakdown

### `s = str(x)`

`str()` converts the integer into a **string** (text). This lets us treat the number as a sequence of characters that we can reverse and compare.

Why? Because integers don't have a built-in concept of "reading backwards." But strings do — they're just sequences of characters, and Python makes it easy to reverse sequences.

````python
# str() converts any type to its string representation
print(str(121))     # '121'   — 3 characters: '1', '2', '1'
print(str(-121))    # '-121'  — 4 characters: '-', '1', '2', '1'
print(str(10))      # '10'   — 2 characters: '1', '0'
print(str(0))       # '0'    — 1 character: '0'

# The type changes from int to str
print(type(121))       # <class 'int'>
print(type(str(121)))  # <class 'str'>
````

### `s[::-1]` — Reversing a String

This is Python's **slice notation**. The full syntax is `s[start:stop:step]`.

- `start` — where to begin (default: beginning)
- `stop` — where to end (default: end)
- `step` — how many characters to skip (default: 1)

When `step` is **-1**, Python walks through the string **backwards**, effectively reversing it.

`s[::-1]` means: "start at the end, go to the beginning, one character at a time in reverse."

````python
s = "hello"

# Normal slicing
print(s[0:3])     # 'hel'  — characters at index 0, 1, 2
print(s[1:4])     # 'ell'  — characters at index 1, 2, 3
print(s[::2])     # 'hlo'  — every 2nd character (step = 2)

# Reverse slicing
print(s[::-1])    # 'olleh' — the whole string, backwards

print(s[::1])     # 'hello' - the whole string, not backwards

# Step by step, what s[::-1] does:
# s[4] = 'o'
# s[3] = 'l'
# s[2] = 'l'
# s[1] = 'e'
# s[0] = 'h'
# Result: 'olleh'
````

Let's see how this works on our test cases:

````python
examples = ['121', '-121', '10', '12321', '0', '1001']

for s in examples:
    rev = s[::-1]
    match = "✅ palindrome" if s == rev else "❌ not palindrome"
    print(f"'{s}'  reversed →  '{rev}'  {match}")
````

### `return s == s[::-1]`

The `==` operator compares the original string with the reversed string, character by character:

- If every character matches → returns `True` (it's a palindrome)
- If any character differs → returns `False` (not a palindrome)

This single line handles all cases, including negatives and trailing zeros, because:
- `-121` reversed is `121-` → characters don't match → `False`
- `10` reversed is `01` → characters don't match → `False`
- `121` reversed is `121` → all characters match → `True`

---
## Deep Dive: How String Comparison Works

When Python compares two strings with `==`, it checks each character at the same position:

````python
s = '121'
rev = s[::-1]  # '121'

print("Comparing character by character:")
for i in range(len(s)):
    match = "✅" if s[i] == rev[i] else "❌"
    print(f"  Position {i}: s[{i}] = '{s[i]}'  vs  rev[{i}] = '{rev[i]}'  {match}")

print(f"\nAll match? {s == rev}")

print("\n" + "="*40)

s = '-121'
rev = s[::-1]  # '121-'

print(f"\nComparing '{s}' vs '{rev}':")
for i in range(len(s)):
    match = "✅" if s[i] == rev[i] else "❌"
    print(f"  Position {i}: s[{i}] = '{s[i]}'  vs  rev[{i}] = '{rev[i]}'  {match}")

print(f"\nAll match? {s == rev}")
````

## Edge Cases

Always think about edge cases — inputs that might break your logic or behave unexpectedly:

````python
sol = Solution()

edge_cases = [
    (0, "zero"),
    (5, "single digit"),
    (-1, "negative single digit"),
    (11, "two identical digits"),
    (10, "trailing zero"),
    (100, "multiple trailing zeros"),
    (1000001, "large palindrome"),
    (1000000, "large non-palindrome"),
]

for x, description in edge_cases:
    result = sol.isPalindrome(x)
    print(f"x = {x:>8}  ({description:.<30})  →  {result}")
````

## Complexity

- **Time: O(n)** where n is the number of digits. `str(x)` is O(n), `s[::-1]` is O(n), and `==` comparison is O(n). Three O(n) operations = still O(n).
- **Space: O(n)** because we create two strings: `s` and `s[::-1]`, each with n characters.

This is efficient enough for the given constraints (-2³¹ to 2³¹ - 1), which means at most 10-11 digits.

Three Python built-in operations, one line of actual logic. The negative sign and trailing zeros are handled automatically without any special cases.
