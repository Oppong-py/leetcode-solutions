# LeetCode #1 — Two Sum

Difficulty: Easy | [Problem Link](https://leetcode.com/problems/two-sum/)

## The Problem (Restated Simply)

You have a list of integers and a target number. Find the **two positions (indices)** of numbers in the list that add up to the target. There is always exactly one valid answer.

````python
nums = [2, 7, 11, 15]
target = 9
# Answer: [0, 1] because nums[0] + nums[1] = 2 + 7 = 9
````

## The Starter Code

LeetCode gives you this to complete:

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
```

### Breaking down that definition line:

- **`class Solution:`** — LeetCode wraps everything in a class. You don't need to worry about this.
- **`def twoSum`** — Defining a method (function inside a class) called `twoSum`.
- **`self`** — Required as the first parameter in every method inside a class. You won't use it — it's just Python's formality. It's not exclusive to `__init__`; every class method needs it.
- **`nums: List[int]`** — The parameter `nums` with a *type hint* saying it should be a list of integers. The colon here is not starting a code block — it separates the name from the type. Type hints are purely informational; Python doesn't enforce them.
- **`target: int`** — The parameter `target` with a type hint saying it should be an integer.
- **`-> List[int]`** — A *return type hint* saying this function returns a list of integers.

Stripped of all hints, the function is just `def twoSum(self, nums, target):` — same behavior.

# Approach 1: Brute Force

## The Idea

Check **every possible pair** of numbers in the list. If any pair adds up to the target, return their indices.

This is the most straightforward approach — no tricks, no data structures, just two nested loops.

## The Full Solution

````python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]

# Quick test
sol = Solution()
print(sol.twoSum([2, 7, 11, 15], 9))   # [0, 1]
print(sol.twoSum([3, 2, 4], 6))         # [1, 2]
print(sol.twoSum([3, 3], 6))            # [0, 1]
````

## Line-by-Line Breakdown

### `for i in range(len(nums)):`

We don't know how many items are in the list — it could be any length. `len(nums)` gives us the count, and `range()` creates a sequence from `0` to `len(nums) - 1`. So `i` visits every index in the list, one at a time.

### `for j in range(i + 1, len(nums)):`

For each `i`, we loop through every index **after** it. Starting from `i + 1` ensures two things:
- We never compare a number with itself
- We never check the same pair twice (e.g., if we checked `(0, 1)`, we won't also check `(1, 0)`)

### `if nums[i] + nums[j] == target:`

Simple check — do the two numbers at positions `i` and `j` add up to the target?

### `return [i, j]`

If yes, return both indices as a list. The function exits immediately.

## Complexity

- **Time: O(n²)** — For each of the n elements, we check up to n-1 others. For a list of 10,000 items, that's up to ~50 million checks.
- **Space: O(1)** — No extra data structures. Just two loop variables.

This works and is correct, but it's slow for large inputs. That's where Approach 2 comes in.

# Approach 2: Hash Map (Optimal)

## The Idea

Instead of checking every pair, use a **dictionary** to remember numbers you've already seen. For each number, calculate what its partner would need to be (`target - num`), and check if that partner is already in the dictionary. This turns the inner loop into an instant lookup.

## The Full Solution

````python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i

# Quick test
sol = Solution()
print(sol.twoSum([2, 7, 11, 15], 9))   # [0, 1]
print(sol.twoSum([3, 2, 4], 6))         # [1, 2]
print(sol.twoSum([3, 3], 6))            # [0, 1]
````

## Line-by-Line Breakdown

### Line 1: `seen = {}`

This creates an **empty dictionary** (also called a hash map — same thing in Python).

A dictionary is a lookup table that stores **key-value pairs**. Think of it like a contact list on your phone: the name is the key, the phone number is the value. You look up a name and instantly get the number.

In this problem:
- **Key** = a number from the list
- **Value** = the index (position) where that number sits in the list

As we walk through the list, we store each number we've visited in this dictionary. That way, future iterations can check: "has the number I need already appeared?"

Why is this fast? Checking if a key exists in a dictionary is **O(1)** — constant time, effectively instant — no matter how big the dictionary is. This is what makes the hash map approach faster than brute force.

````python
# Example of how the dictionary grows:
seen = {}
print(seen)              # {}

seen[2] = 0
print(seen)              # {2: 0}

seen[7] = 1
print(seen)              # {2: 0, 7: 1}

seen[11] = 2
print(seen)              # {2: 0, 7: 1, 11: 2}
````

---
### Line 2: `for i, num in enumerate(nums):`

This loops through the list, giving us both the **index** and the **value** at each step.

---
## Deep Dive: What is `enumerate`?

`enumerate()` is a built-in Python function that wraps around any iterable (like a list) and gives you back **pairs** of (index, value) on each iteration.

### Without enumerate — the manual way:

````python
nums = [2, 7, 11, 15]

# Without enumerate — manual indexing
for i in range(len(nums)):
    num = nums[i]
    print(f"Index: {i}, Value: {num}")
````

### With enumerate — the Pythonic way:

````python
nums = [2, 7, 11, 15]

# With enumerate — cleaner, same result
for i, num in enumerate(nums):
    print(f"Index: {i}, Value: {num}")
````

### What `enumerate` actually returns

Under the hood, `enumerate` produces **tuples** (pairs):

````python
nums = [2, 7, 11, 15]
print(list(enumerate(nums)))
# Each tuple is (index, value)
````

When you write `for i, num in enumerate(nums)`, Python **unpacks** each tuple into two separate variables: `i` gets the first element (the index) and `num` gets the second (the value).

This unpacking is the same concept as:

````python
# Tuple unpacking
pair = (0, 2)
i, num = pair
print(f"i = {i}, num = {num}")

# It works with any number of variables
a, b, c = (10, 20, 30)
print(f"a = {a}, b = {b}, c = {c}")
````

### Optional: custom start index

By default, `enumerate` starts counting from 0. You can change this:

````python
nums = [2, 7, 11, 15]

for i, num in enumerate(nums, start=1):
    print(f"Position {i}: {num}")

# We don't use start= in this problem because list indices begin at 0
````

### Why use `enumerate` over `range(len(...))`?

- **Readability** — `for i, num in enumerate(nums)` immediately tells you "I need both the index and the value." With `range(len(nums))`, the reader has to look for where `nums[i]` appears.
- **Less error-prone** — You can't accidentally write `nums[j]` when you meant `nums[i]` because the value is already in `num`.
- **Pythonic convention** — If you see `for i in range(len(something))` in a Python code review, most experienced developers will suggest using `enumerate` instead.

---
### Line 3: `complement = target - num`

This is the **key insight** of the entire problem.

If two numbers must add up to `target`, and you're currently looking at `num`, then the other number you need is `target - num`. We call this the **complement**.

````python
# Complement examples
examples = [(9, 2), (9, 7), (6, 3), (6, 2)]

for target, num in examples:
    complement = target - num
    print(f"target = {target}, num = {num}  →  complement = {target} - {num} = {complement}  (you need a {complement})")
````

---
### Line 4: `if complement in seen:`

Check the dictionary: have we already encountered the complement in a previous iteration?

- `in` checks if a **key** exists in the dictionary. This is O(1) — instant.
- If yes → we found our pair. The complement's index is stored in the dictionary, and the current number's index is `i`.
- If no → move on, store the current number, and keep looking.

---
### Line 5: `return [seen[complement], i]`

We found the answer. Build and return a list of two indices:

- `seen[complement]` — look up the complement in the dictionary to get its index (the index where we saw that number earlier)
- `i` — the current index

The earlier index comes first because we encountered it in a previous iteration.

---
### Line 6: `seen[num] = i`

If we didn't find the complement, store the current number in the dictionary so **future iterations** can find it.

This line only runs when the `if` on line 4 is `False` (i.e., the complement wasn't found yet). We're essentially saying: "I haven't found my partner yet, but let me register myself in case someone later needs me."

---
## Full Walkthrough: Example 1

`nums = [2, 7, 11, 15]`, `target = 9`

````python
# Walkthrough Example 1: step by step with print statements
nums = [2, 7, 11, 15]
target = 9
seen = {}

for i, num in enumerate(nums):
    complement = target - num
    print(f"Iteration {i+1}: i={i}, num={num}")
    print(f"  complement = {target} - {num} = {complement}")
    
    if complement in seen:
        print(f"  Is {complement} in seen ({seen})?  ✅ YES — at index {seen[complement]}")
        result = [seen[complement], i]
        print(f"  Return {result}")
        break
    else:
        print(f"  Is {complement} in seen ({seen})?  ❌ No")
        seen[num] = i
        print(f"  Store → seen = {seen}")
    print()
````

---
## Full Walkthrough: Example 3

`nums = [3, 3]`, `target = 6`

Notice how this handles duplicate values correctly. The first `3` gets stored in the dictionary, and when the second `3` comes along, it finds its complement (which is also `3`) already in the dictionary. The order of operations matters — we check *before* we store, so a number never matches with itself.

````python
# Walkthrough Example 3 — duplicate values
nums = [3, 3]
target = 6
seen = {}

for i, num in enumerate(nums):
    complement = target - num
    print(f"Iteration {i+1}: i={i}, num={num}")
    print(f"  complement = {target} - {num} = {complement}")
    
    if complement in seen:
        print(f"  Is {complement} in seen ({seen})?  ✅ YES — at index {seen[complement]}")
        result = [seen[complement], i]
        print(f"  Return {result}")
        break
    else:
        print(f"  Is {complement} in seen ({seen})?  ❌ No")
        seen[num] = i
        print(f"  Store → seen = {seen}")
    print()
````

# Comparing Both Approaches

| | Brute Force (Approach 1) | Hash Map (Approach 2) |
|---|---|---|
| **Time** | O(n²) — check every pair | O(n) — one pass through the list |
| **Space** | O(1) — no extra memory | O(n) — dictionary stores up to n items |
| **10,000 items** | Up to 50,000,000 checks | At most 10,000 checks |
| **Readability** | Very intuitive | Requires understanding dictionaries |
| **Interview** | Shows you can solve it | Shows you can optimize |

You trade a bit of memory (the dictionary) for a massive speed improvement. This tradeoff — using space to save time — is one of the most fundamental patterns in computer science and comes up constantly in LeetCode problems.

---
## Summary

1. Walk through the list one number at a time
2. For each number, calculate what its partner would need to be (`target - num`)
3. Check if that partner already exists in the dictionary (instant lookup)
4. If yes → return both indices. Problem solved.
5. If no → store the current number in the dictionary and move on

The dictionary acts as your memory of everything you've already seen, turning a "search the whole list" operation into an instant lookup.
